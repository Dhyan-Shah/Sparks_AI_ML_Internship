# Basics of Seaborn Visualization Library

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

#Create a scatterplot: bill_length_mm vs bill_depth_mm with title and custom color.
df=pd.read_csv('penguins.csv')

sns.scatterplot(data=df, x='bill_length_mm', y='bill_depth_mm', color='orange')
plt.title('Bill Length vs Bill Depth of Penguins')
plt.show()
```

```python
#Scatterplot: bill_length_mm vs flipper_length_mm using hue='species' and palette='Set2'.
sns.scatterplot(data=df, x='bill_length_mm', y='flipper_length_mm', hue='species', palette='Set2')
plt.title('Bill Length vs Flipper Length of Penguins by Species')
plt.show()
```

```python
#Scatterplot: bill_length_mm vs flipper_length_mm using hue='species' and palette='Set2'.
sns.scatterplot(data=df, x='bill_length_mm', y='flipper_length_mm', hue='species', palette='Set2')
plt.title('Bill Length vs Flipper Length of Penguins by Species')
plt.show()
```

```python
#Scatterplot: bill_length_mm vs body_mass_g using hue='species' and style='sex'.
sns.scatterplot(data=df, x='bill_length_mm', y='body_mass_g', hue='species', style='sex')
plt.title('Bill Length vs Body Mass of Penguins by Species and Sex')
plt.show()
```

```python
#Create lineplot using bill_length_mm vs body_mass_g with hue='species' and marker='o'.
sns.lineplot(data=df, x='bill_length_mm', y='body_mass_g', hue='species', marker='o')
plt.title('Bill Length vs Body Mass of Penguins by Species')
plt.show()
```

```python
#Histogram of body_mass_g using bins=20, kde=True.
sns.histplot(data=df, x='body_mass_g', bins=20, kde=True)
plt.title('Distribution of Body Mass of Penguins')
plt.show()
```

```python
#Histogram of body_mass_g using hue='species', bins=15, kde=True.
sns.histplot(data=df, x='body_mass_g', hue='species', bins=15, kde=True)
plt.title('Distribution of Body Mass of Penguins by Species')
plt.show()
```

```python
#Create heatmap using annot=True and cmap='coolwarm'.
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap of Penguin Dataset')
plt.show()
```
