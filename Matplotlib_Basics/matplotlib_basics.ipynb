```python
#Q1
#Create a line plot with title, x-label, y-label
import numpy as np
import matplotlib.pyplot as plt

x= np.linspace(0, 10, 100)
y=np.linspace(0, 20, 100)
plt.plot(x, y)
plt.title("Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
#Q2
#Plot two lines on same graph with legend and colors.
x= np.linspace(0, 10, 100)
y=np.linspace(0, 20, 100)
plt.plot(x, y, label="Line 1", color='blue')
y2=np.linspace(0, 30, 100)
plt.plot(x, y2, label="Line 2", color='red')
plt.title("Two Lines Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend()
plt.show()
```

```python
#Q3
#Create a scatter plot and customize marker, color, size
x= np.random.rand(100)
y= np.random.rand(100)
plt.scatter(x, y, marker='o', color='hotpink', s=100, alpha=0.5)
x= np.random.rand(100)
y= np.random.rand(100)
plt.scatter(x, y, marker='x', color='cyan', s=100, alpha=0.5)
plt.title("Scatter Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
#Q4
#Create a bar chart for student marks with custom colors.
students = ['Alice', 'Bob', 'Charlie', 'David', 'Eve']
marks = [85, 90, 78, 92, 88]
colors = ['red', 'blue', 'green', 'orange', 'purple']
plt.bar(students, marks, color=colors)
plt.title("Student Marks")

plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

```python
#Q5
#Create a pie chart with percentage labels, shadow, explode
labels = ['Python', 'JavaScript', 'Java', 'C++', 'Ruby']
sizes = [30, 25, 20, 15, 10]
explode = (0.1, 0, 0, 0, 0)
plt.pie(sizes, explode=explode, labels=labels, shadow=True)
```

```python
#Q6
#Create a line plot using x=np.arange(1,11), y=x**2 with grid, linestyle, marker
x = np.arange(1, 11)
y = x**2    
plt.minorticks_on()
plt.grid(which='major', linestyle='-', linewidth='0.5', color='red')
plt.grid(which='minor', linestyle=':', linewidth='0.5', color='blue')
plt.plot(x, y, linestyle='--', marker='o', color='green')
plt.title("Line Plot with Grid")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
#Q7
#Plot sine and cosine curves on same graph with legend.
x = np.linspace(0, 2 * np.pi, 100)
y1 = np.sin(x)

y2 = np.cos(x)
plt.plot(x, y1, label='Sine', color='blue')
plt.plot(x, y2, label='Cosine', color='orange')
plt.title("Sine and Cosine Curves")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend()
plt.show()
```

```python
#Q8
#Create a histogram of 100 random numbers with 10 bins.
data = np.random.randn(100)
plt.hist(data, bins=10, color='purple', alpha=0.7)
plt.title("Histogram of Random Numbers")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```

```python
#Q9
#Generate 50 random marks, create histogram and average line
marks = np.random.randint(0, 101, 50)
plt.hist(marks, bins=10, color='cyan', alpha=0.7)
average = np.mean(marks)
plt.axvline(average, color='red', linestyle='dashed', linewidth=2, label=f'Average: {average:.2f}')
plt.title("Histogram of Marks with Average Line")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

```python
##Q11
#Create 1x2 subplots: line plot + bar chart
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
x = np.linspace(0, 10, 100)
y = np.sin(x)
axes[0].plot(x, y, color='blue')
axes[0].set_title("Sine Curve")
axes[0].set_xlabel("X-axis")
axes[0].set_ylabel("Y-axis")
students = ['Alice', 'Bob', 'Charlie', 'David', 'Eve']
marks = [85, 90, 78, 92, 88]
colors = ['red', 'blue', 'green', 'orange', 'purple']
axes[1].bar(students, marks, color=colors)
axes[1].set_title("Student Marks")
axes[1].set_xlabel("Students")
axes[1].set_ylabel("Marks")
plt.tight_layout()
plt.show()
```

```python
#Q12
#Create 2x2 subplot dashboard with line, scatter, hist, pie
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

plt.suptitle("Matplotlib Dashboard", fontsize=16)

x = np.linspace(0, 10, 100)
y = np.sin(x)
axes[0, 0].plot(x, y, color='blue')
axes[0, 0].set_title("Line Plot")
axes[0, 0].set_xlabel("X-axis")
axes[0, 0].set_ylabel("Y-axis")

x = np.random.rand(100)
y = np.random.rand(100)
axes[0, 1].scatter(x, y, marker='o', color='hotpink', s=200, alpha=0.5)
axes[0, 1].set_title("Scatter Plot")
axes[0, 1].set_xlabel("X-axis")
axes[0, 1].set_ylabel("Y-axis")

data = np.random.randn(100)
axes[1, 0].hist(data, bins=10, color='purple', alpha=0.7)
axes[1, 0].set_title("Histogram")
axes[1, 0].set_xlabel("Value")
axes[1, 0].set_ylabel("Frequency")

labels = ['Python', 'JavaScript', 'Java', 'C++', 'Ruby']
sizes = [35, 25, 20, 15, 12]
explode = (0.1, 0, 0, 0, 0)
axes[1, 1].pie(sizes, explode=explode, labels=labels, shadow=True)
axes[1, 1].set_title("Pie Chart")

plt.tight_layout()
plt.show()
```
