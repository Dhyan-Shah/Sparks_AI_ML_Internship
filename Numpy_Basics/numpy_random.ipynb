```python
#Q1.

#Generate an array of 5 random float numbers between 0 and 1 using np.random.rand().
import numpy as np
arr=np.random.rand(5)
print(arr)
```

```python
#Q6.

#Use np.random.choice() to select 3 random elements from the array:

arr = np.array([10, 20, 30, 40, 50])
choice=np.random.choice(arr,3)
print(choice)
```

```python
#Q7.

#Use np.random.choice() to pick 5 elements with replacement from:

arr = np.array([1, 2, 3, 4, 5])
choice = np.random.choice(arr,5,replace=True)
print(choice)
```

```python
#Q8.

#Shuffle this array using np.random.shuffle():

arr = np.array([1, 2, 3, 4, 5])
np.random.shuffle(arr)
print(arr)
```

```python
#Q10.

#Generate a 4x4 matrix of random integers between 10 and 99 and find:
arr=np.random.randint(10,99,size=(4,4))
print(arr)

#Maximum value
print("Maximum value:",np.max(arr))

#Minimum value
print("Minimum value:",np.min(arr))

#Mean
print("Mean:",np.mean(arr))
```

```python
#Q11. Student Marks Simulation 

#Simulate marks (0–100) of 15 students using np.random.randint()
marks = np.random.randint(0, 101, size=15)
print("Marks of 15 students:", marks)

#Find the top 5 marks (sorted descending).
print("Top 5 marks:", np.sort(marks)[-5:][::-1])

#Randomly select 3 students for a prize.
selected_students = np.random.choice(15, 3, replace=False)
print("Selected students (indices):", selected_students)
```

```python
#Generate daily sales for 7 days between 200 and 1000 units.

sales = np.random.randint(200, 1001, size=7)
print("Daily sales for 7 days:", sales)

#Find the day with the highest sales.
print("Day with highest sales (index):", np.argmax(sales))
print("Highest sales value:", np.max(sales))

#Randomly pick 2 days for a special offer.
print("Randomly picked days for special offer (indices):", np.random.choice(7, 2, replace=False))


#Shuffle the sales data.
np.random.shuffle(sales)
print("Shuffled sales data:", sales)
```

```python
#Q13. Employee Salary Data 

#Create salary data for 8 employees between 30,000 and 80,000.
arr = np.random.randint(30000, 80001, size=8)
print("Employee salaries:", arr)

#Find the employee with the lowest salary.
print("Lowest salary:", np.min(arr))

#Randomly pick 2 employees for a bonus.
bonus_employees = np.random.choice(8, 2, replace=False)
print("Employees selected for bonus (indices):", bonus_employees)

```

```python
#Q14. Product Prices 

#You have 10 products. Generate random prices between ₹100 and ₹1000.

prices = np.random.randint(100, 1001, size=10)
print("Product prices:", prices)

#Sort the prices in ascending order.
sorted_prices = np.sort(prices)
print("Sorted prices:", sorted_prices)

#Pick 3 random products for a discount.
discount_products = np.random.choice(10, 3, replace=False)
print("Products selected for discount (indices):", discount_products)
```

```python
#Q15. Weather Temperature Data 

#Simulate 10 random daily temperatures between 25°C and 45°C.
temperatures = np.random.randint(25, 46, size=10)
print("Daily temperatures:", temperatures)

#Find 3 hottest days using argsort().
hottest_days = np.sort(temperatures)[-3:]
print("Hottest days:", hottest_days)

#Randomly shuffle the temperature data.
np.random.shuffle(temperatures)
print("Shuffled temperatures:", temperatures)
```

```python
#Q16. Exam Simulation 

#You are simulating marks of 1000 students (0–100).
arr = np.random.randint(0, 101, size=1000)
print("Marks of 1000 students:", arr)

#Find the average mark.
print("Average mark:", np.mean(arr))

#Find how many scored above 80.
above_80 = np.sum(arr > 80)
print("Number of students who scored above 80:", above_80)

#Randomly select 10 for interview.
interviewees = np.random.choice(1000, 10, replace=False)
print("Students selected for interview (indices):", interviewees)
```

```python
#Q17. Random Sampling in Population 

#Create an array representing 1000 people IDs (1–1000).
arr = np.arange(1, 1001)
# print("People IDs:", arr)

#Randomly select 50 people for a survey (no repetition).
survey_participants = np.random.choice(arr, 50, replace=False)
print("Survey participants (IDs):", survey_participants)

#Randomly select 5 people for a prize (with repetition).
prize_winners = np.random.choice(arr, 5, replace=True)
print("Prize winners (IDs):", prize_winners)
```

```python
#Q18. Game Simulation 

#A dice game is played 10 times.
game_results = np.random.randint(1, 7, 10)
print("Dice game results:", game_results)

#Simulate the results using np.random.randint(1, 7, 10).

#Find how many times 6 appeared.
six_count = np.sum(game_results == 6)
print("Number of times 6 appeared:", six_count)
```
