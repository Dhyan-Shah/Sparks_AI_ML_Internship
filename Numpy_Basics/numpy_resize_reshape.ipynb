```python
# Q-1: Create a 1D NumPy array of size 10.
import numpy as np 
arr=np.arange(10)
#->Find its shape,ndim and size.

print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
# Q-2: Make a 2D array of shape(3,4) with values from 0 to 11.
arr= np.arange(12).reshape(3,4)

# ->Print its shape,ndim and size.
print(arr.shape)
print(arr.ndim)
print(arr.size)
```

```python
#Q-3: Convert a 1D array of size 12 into:
arr=np.arange(12)

#->shape(3,4)
print(arr.reshape(3,4))

#->shape(2,2,3)
print(arr.reshape(2,2,3))

#->shape(6,2)
print(arr.reshape(6,2))
```

```python
#Q-4: Use -1 in reshape to automatically calculate the missing dimension. Example reshape(3,4) into (2,-1)

print(arr.reshape(2,-1))
```

```python
#Q-5: Create a 2D array of shape (3,3) and:
arr=np.random.randint(1,10,size=(3,3))
print(arr)

#->Flatten it into 1D using flatten()
print(arr.flatten())

#->Convert it into 1D using ravel() and change one element.
raveled=arr.ravel()
raveled[0]=100
print(raveled)

#->Check whether original array changes or not.
print(arr)
```

```python
#Q-6 Take a 1D array [1,2,3,4]
arr=np.array([1,2,3,4]) 

#->Convert it into shape(1,4) using np.newaxis
arr_reshaped = arr[np.newaxis, :]
print(arr_reshaped)

#->Convert it into shape(4,1) using np.expand_dims
arr_expanded = np.expand_dims(arr, axis=1)
print(arr_expanded)
```

```python
#Q-7: Create an array of shape(1,5,1) and remove all extra dimensions using squeeze().
arr=np.array([[[1],[2],[3],[4],[5]]])
print(arr)
arr_squeezed=np.squeeze(arr)
print(arr_squeezed)
```

```python
#Q-8: Make a 2D array of shape(2,3) and print its transpose.
arr=np.array([[1,2,3],[4,5,6]])
print(arr)
print(arr.T)

#->Verify shape before and after transpose.
print("Original shape:", arr.shape)
print("Transposed shape:", arr.T.shape)
```

```python
#Q-9: Create a 3D array of shape (2,3,4) and use transpose() to swap axes.
arr=np.random.randint(1,10,size=(2,3,4))
print("Original array shape:", arr.shape)

#->Try arr.transpose(1,0,2) and check new shape.
arr_transposed = arr.transpose(1,0,2)
print("Transposed array shape:", arr_transposed.shape)
```

```python
#Q-11:You have a dataset with 1000 rows (samples) and 10 features->shape(1000,10).


#(a) Find its ndim,size and shape.
arr=np.random.rand(1000,10)
print(arr.ndim)
print(arr.size)
print(arr.shape)

#(b) Reshape it into (10,1000) and observe the diffrence.
arr_reshaped = arr.reshape(10,1000)
print(arr_reshaped.shape)

#(c) Extract a single feature as a 1D vector (1000,).
feature_1d = arr[:, 0]  # Extract the first feature
print(feature_1d.shape)

#(d) Extract a single feature as a column vector (1000,1).
feature_column = arr[:, 0:1]  # Extract the first feature as a column vector
print(feature_column.shape)
```

```python
#Q-14:A dataset has shape(10000,20) (10,000 samples,20 features).
arr=np.random.rand(10000,20)

#(a)Reshape it into (100,100,20)->100 batches,each with 100 samples.
arr_batches = arr.reshape(100, 100, 20)
print(arr_batches.shape)

#(b)Select one batch(100,20) and transpose it into (20,100).
batch = arr_batches[0]  
batch_transposed = batch.transpose(1, 0)
print(batch_transposed.shape)
```

```python
#Q-15:You have daily sensor data with shape(365,24)->365 days 24 hourly readings each day.

arr=np.random.rand(365,24)

#(a)Flatten it into (8760,).
arr_flattened = arr.flatten()
print(arr_flattened.shape)

#(b)Add an extra dimension to make (365,24,1) for ML input.
arr_3d = arr.reshape(365, 24, 1)
print(arr_3d.shape)

#(c)Use sequeeze to bring it back to (365,24).
arr_squeezed = arr_3d.squeeze()
print(arr_squeezed.shape)
```

```python
#Q-16: You have stock prices with shape(30,5)->30 days, 5 companies.

arr=np.random.rand(30,5)

#(a)Extract one company's stock price as a 1D array(30,).
print(arr[0, :])

#(b)Convert the same into a column vector (30,1).
arr_column = arr[:, 0].reshape(-1, 1)
print(arr_column.shape)

#(c)Transpose the dataset to (5,30)->stocks*days.
arr_transposed = arr.T
print(arr_transposed.shape)
```

```python
#Q-18:You have 200 senetences,each represented by 50 words,each word embedding of 100 dimensions.
arr=np.random.rand(200,50,100)
#->shape(200,50,100)

#(a)Flatten each sentence into (200,5000) for a bag-of-word style model.
arr_flattened = arr.reshape(200, -1)
print(arr_flattened.shape)

#(b)Transpose it to (200,100,50) so embeddings come before word positions.
arr_transposed = arr.transpose(0, 2, 1)
print(arr_transposed.shape)

#(c)Extract one sentence and reshape into (50,100)again.
sentence = arr[0]
sentence_reshaped = sentence.reshape(50, 100)
print(sentence_reshaped.shape)
```

```python
#Task 10

#Using nested for-loops, print each element of the 3D array.
arr=np.arange(12).reshape(2,2,3)
for i in range(arr.shape[0]):
    for j in range(arr.shape[1]):
        for k in range(arr.shape[2]):
            print(arr[i, j, k])
```

```python
#Task 11

#Using np.nditer():

#Print all elements of a 3D array in one line

print("Elements of the 3D array in one line:")
for element in np.nditer(arr):
    print(element, end=' ')
#Count total elements manually
total_elements = 0
for element in np.nditer(arr):
    total_elements += 1
print(f"\nTotal elements: {total_elements}")
```

```python
#Task 12

#Using np.ndenumerate():


#Print index and value like this:

print("Index and value of each element:")
for index, value in np.ndenumerate(arr):
    print(f"Index: {index} Value: {value}")
#Index: (0, 1, 2) Value: 5
```

```python
#Task 13 (Thinking Task ⭐)

#From a 3D array:

#Print only elements greater than 5
arr=np.arange(12).reshape(2,2,3)
print("Elements greater than 5:")
for index, value in np.ndenumerate(arr):
    if value > 5:
        print(f"Index: {index} Value: {value}")
# Also print their index positions
```
