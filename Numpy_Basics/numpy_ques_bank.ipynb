```python
#26. What is the fundamental difference between flattening an array using flatten() versus 
#using ravel()? 
# The fundamental difference between flatten() and ravel() is that flatten() returns a copy of the array, while ravel() returns a view of the original array whenever possible. This means that changes made to the array returned by flatten() will not affect the original array, whereas changes made to the array returned by ravel() may affect the original array if it is a view.
arr=np.array([[1,2,3],[4,5,6]])
flattened_arr = arr.flatten()
print(flattened_arr)

raveled_arr = arr.ravel()
print(raveled_arr)
```

```python
#27. How do you add a new axis to convert a 1D array [1, 2, 3, 4] into a column vector of 
#shape (4, 1) using np.expand_dims()? 
arr=np.array([1,2,3,4])
new_arr=np.expand_dims(arr,axis=1)
print(new_arr)
```

```python
#28. Write code to remove all extra dimensions of size 1 from an array of shape (1, 5, 1) 
#using the np.squeeze() function. 
arr=np.array([[[1],[2],[3],[4],[5]]])
print("Original shape:", arr.shape)
squeezed_arr=np.squeeze(arr)
print("Squeezed shape:", squeezed_arr.shape)
```

```python
#29. How do you swap the axes (transpose) of a 2D array? 
arr=np.array([[1,2,3],[4,5,6]])
transposed_arr=arr.T
print(transposed_arr)
```

## File: N-4_Iterating.pdf (Iterating Arrays)


```python
#31. Write a standard Python for loop to iterate through all elements of a basic 1D array. 
arr=np.array([10,20,30,40,50])
for i in arr:
    print(i)
```

```python
#32. How do you iterate through a 2D array element-by-element using nested loops? 
arr=np.arange(1,10).reshape(3,3)
for i in arr:
    for j in i:
        print(j)
```

```python
#33. Write the code to iterate through every element of a 3D array efficiently using 
#np.nditer()
arr=np.random.randint(0,10,(2,2,3))
for x in np.nditer(arr):
    print(x)
```

```python
#34. How do you iterate through an array while keeping track of the exact index of each 
#element using np.ndenumerate()?

arr=np.array([[11,12,13],[21,22,23],[31,32,33]])
for index,value in np.ndenumerate(arr):
    print(f"Index: {index}, Value: {value}")
```

```python
#35. Write code to modify a 2D array in-place during iteration by replacing all elements at 
#even row indices with 0.

arr = np.array([[11, 12, 13], [21, 22, 23], [31, 32, 33]])
for i in range(arr.shape[0]):
    if i % 2 == 0:  
        arr[i, :] = 0  
print(arr)
```

```python
#36. Given two arrays a = np.array([1, 2, 3]) and b = np.array([[10], [20], [30]]), write code 
#to print their element-wise sum using broadcasting with np.nditer([a, b]).
a=np.array([1,2,3])
b=np.array([[10],[20],[30]])
for x,y in np.nditer([a,b]):
    print(x+y)
```

```python
#37. How do you flatten an N-dimensional array first so that you can iterate over it as a 
#standard 1D array in a single loop? 
arr=np.random.randint(0, 10, (2, 2, 3))
for x in arr.flatten():
    print(x)
```

```python
#47. Given a 2D array representing quarterly revenue for 3 years, write the code to split this 
#data into 4 parts, representing Q1, Q2, Q3, and Q4.
revenue_data = np.array([[100, 150, 200, 250],
                        [300, 350, 400, 450],
                        [500, 550, 600, 650]])
q1_data, q2_data, q3_data, q4_data = np.hsplit(revenue_data, 4)
print("Q1 Data:\n", q1_data)
print("Q2 Data:\n", q2_data)
print("Q3 Data:\n", q3_data)
print("Q4 Data:\n", q4_data)
```
